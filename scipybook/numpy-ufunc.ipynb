# ブロードキャストとufunc ― 配列の要素ごとの計算

## この章で学ぶこと

ufunc は universal function の略で、配列の要素ごとに計算する関数です。NumPy に組み込まれた ufunc の多くは C 言語で書かれているので、Python のループよりずっと速く計算できます。この章では、ufunc の使い方と、形の違う配列どうしを計算するブロードキャストを確認します。

- ufunc の基本的な使い方と `out` パラメータ
- 四則演算・比較演算・論理演算に対応する ufunc
- ブロードキャストの 4 つのルール
- `ogrid` と `mgrid` で格子を作る方法
- `reduce()`、`accumulate()`、`outer()` といった ufunc のメソッド

まず、ufunc の動きを見てみます。`linspace()` で $0$ から $2\pi$ までを等分した配列を作り、`np.sin()` に渡します。

````python
import helper.matplotlib
import helper.magics
import numpy as np

x = np.linspace(0, 2 * np.pi, 10)
y = np.sin(x)
y
````

`np.sin()` は ufunc なので、配列 `x` の各要素について正弦値を計算し、結果を並べた新しい配列を返します。計算のあとも `x` の値は変わりません。

結果を新しい配列ではなく、既にある配列に書き込みたいときは `out` パラメータを使います。次のコードは、計算結果を `x` 自身に書き込みます。

````python
t = np.sin(x, out=x)   # 結果を x に書き込む
t is x      # True：out=x で x 自身に書き込んだ
````

戻り値は計算結果、つまり `x` そのものです。`np.sin()` は単独の数値にも使えますが、1 つの数値の計算では標準ライブラリ `math` の `sin()` よりかなり遅くなります。`np.sin()` は配列と単独の数値の両方を受け取れるように、C 言語の実装が複雑になっているためです。

戻り値の型も違います。`math.sin()` は Python の `float` を返し、`np.sin()` は `float64` を返します。

````python
import math

%C type(math.sin(0.5)); type(np.sin(0.5))
````

## 四則演算

NumPy には多くの ufunc があります。2 つの配列の和を計算する `add()` を例にします。

````python
a = np.arange(0, 4)
b = np.arange(1, 5)
np.add(a, b)                # 要素ごとの和
````

`add()` は、2 つの配列の同じ位置の要素を足した配列を返します。`out` パラメータを指定しない場合は新しい配列が作られ、指定した場合はその配列に結果が書き込まれます。

````python
np.add(a, b, out=a)   # 結果を a に書き込む
a
````

NumPy の配列は演算子にも対応しているので、`a + b` と書けば `np.add(a, b)` と同じ計算になります。`np.add(a, b, out=a)` は `a += b` と書けます。次の表は、配列の演算子と対応する ufunc の一覧です。

| 式        |  対応するufunc関数      |
|-------------|:-------------------|
|`y = x1 + x2`  |  `add(x1, x2 [, y])` |
|`y = x1 - x2`  |  `subtract(x1, x2 [, y])` |
|`y = x1 * x2`|`multiply(x1, x2 [, y])`|
|`y = x1 / x2`|    `divide(x1, x2 [, y])`|
|`y = x1 // x2`|   `floor_divide(x1, x2 [, y])`|
|`y = -x`      |   `negative(x [,y])`|
|`y = x1**x2`  |   `power(x1, x2 [, y])`|
|`y = x1 % x2` |   `remainder(x1, x2 [, y])`, `mod(x1, x2, [, y])`|

演算子を使うと式が短く書けますが、式が長く配列が大きいと、途中の結果を入れる配列がいくつも作られ、実行速度が落ちます。例えば 3 つの配列 `a`、`b`、`c` で `x = a * b + c` を計算すると、次のようになります。

```python
t = a * b
x = t + c
del t
```

乗算の結果を入れる一時的な配列 `t` が作られ、そのあとで結果の配列 `x` が作られます。メモリの割り当てを 1 回減らすには、次の 2 行に分けます。

```python
x = a * b
x += c
```

## 比較演算と論理演算

`==` や `>` などの比較演算子で 2 つの配列を比べると、各要素を比べた結果のブール配列が返ります。

````python
np.array([1, 2, 3]) < np.array([3, 2, 1])   # 要素ごとに比べる
````

比較演算子にも、対応する ufunc があります。

| 式        |  対応するufunc関数      |
|-------------|:-------------------|
|`y = x1 == x2`|   `equal(x1, x2 [, y]) `|
|`y = x1 != x2`|   `not_equal(x1, x2 [, y])`|
|`y = x1 < x2 `|   `less(x1, x2, [, y]) `|
|`y = x1 <= x2`|   `less_equal(x1, x2, [, y])`|
|`y = x1 > x2 `|   `greater(x1, x2, [, y])`|
|`y = x1 >= x2`|   `greater_equal(x1, x2, [, y])`|

Python の論理演算には `and`、`or`、`not` を使いますが、これらは演算子を上書きできないため、配列の論理演算には対応する ufunc を使います。関数名は `logical_` で始まり、`logical_and`、`logical_not`、`logical_or`、`logical_xor` がそれぞれ論理積、論理否定、論理和、排他的論理和です。次の例は `logical_or()` を使っています。

````python
a = np.arange(5)
b = np.arange(4, -1, -1)
print(a == b)
print(a > b)
print(np.logical_or(a == b, a > b))   # a >= b と同じ
````

2 つのブール配列に `and`、`or`、`not` を使うと `ValueError` になります。ブール配列には `True` と `False` が混ざっているので、NumPy はどちらを返すべきか決められません。

````python
try:
    a == b and a > b
except Exception as ex:
    print(ex)
````

エラーメッセージは、配列の `any()` や `all()` を使うように促します。NumPy の `any()` と `all()` は、Python の組み込み関数と似た働きです。`any()` は `True` が 1 つでもあれば `True` を返し、`all()` はすべての要素が `True` のときだけ `True` を返します。

````python
%C np.any(a == b); np.any(a == b) and np.any(a > b)
````

`bitwise_` で始まる関数はビット演算で、`bitwise_and`、`bitwise_not`、`bitwise_or`、`bitwise_xor` などがあります。`&`、`~`、`|`、`^` の演算子でも同じ計算ができます。

ブール配列では、ビット演算と論理演算の結果は同じです。ただしビット演算子は比較演算子より優先順位が高いので、比較を先に計算するには丸括弧で囲みます。

````python
(a == b) | (a > b)   # 丸括弧がないと a == (b | a) > b と解釈される
````

整数配列のビット演算は C 言語のビット演算と同じです。要素の型の符号に注意してください。`np.arange()` が作る配列の要素型は、NumPy 2 では符号付き 64 ビット整数(int64)です。正数のビットを反転すると負数になります。整数 0 を反転すると `0xFFFFFFFFFFFFFFFF` になり、符号付き 64 ビット整数ではこれが -1 です。

````python
%C ~np.arange(5); np.arange(5).dtype
````

8 ビットの符号なし整数で同じ計算をすると、0 の反転は `0xFF` です。符号なしなので、値は 255 になります。

````python
~np.arange(5, dtype=np.uint8)
````

## ブロードキャスト

ufunc は 2 つの配列の同じ位置の要素を計算するので、配列の形が同じである必要があります。形が違う場合は、次のブロードキャスト(broadcasting)の処理が行われます。

① すべての入力配列を、最も次元の多い配列に合わせる。`shape` 属性の足りない部分は、前に 1 を付けて補う。

② 出力配列の `shape` 属性は、入力配列の `shape` 属性の各軸の最大値になる。

③ 入力配列のある軸の長さが 1 であるか、出力配列のその軸の長さと同じなら、その配列は計算に使える。どちらでもなければエラーになる。

④ 入力配列のある軸の長さが 1 なら、その軸の計算ではその軸の最初の値を使う。

この 4 つのルールは分かりにくいので、実際の例で確かめます。

まず、形状が `(6, 1)` の 2 次元配列 `a` を作ります。

````python
a = np.arange(0, 60, 10).reshape(-1, 1)
%C a; a.shape
````

次に、形状が `(5,)` の 1 次元配列 `b` を作ります。

````python
b = np.arange(0, 5)
%C b; b.shape
````

`a` と `b` の和を計算すると、2 つの配列のすべての要素の組を足した、形状 `(6, 5)` の加算表ができます。

````python
c = a + b
%C c; c.shape
````

`a` と `b` は次元が違うので、ルール①に従って `b` の `shape` を `a` に合わせます。`b` の `shape` の前に 1 を付けて `(1, 5)` にします。NumPy 2.5 では配列の `shape` 属性への代入が非推奨になったので、ここでは `reshape()` を使います。

````python
b = b.reshape(1, 5)
%C b; b.shape
````

これで 2 つの入力の `shape` は `(6, 1)` と `(1, 5)` になりました。ルール②に従うと、出力の各軸の長さは入力の各軸の最大値なので、`(6, 5)` になります。

`b` の第 0 軸の長さは 1、`a` の第 0 軸の長さは 6 です。第 0 軸で足せるようにするには、`b` の第 0 軸を 6 に広げます。

````python
b = b.repeat(6, axis=0)
%C b; b.shape
````

`repeat()` は、`axis` で指定した軸に沿って要素の値を繰り返します。同じように、`a` の第 1 軸の長さは 1、`b` の第 1 軸の長さは 5 なので、`a` の第 1 軸を 5 に広げます。

````python
a = a.repeat(5, axis=1)
%C a; a.shape
````

ここまでの処理で、`a` と `b` は同じ位置の要素を足せるようになりました。ただし `a + b` を計算するとき、NumPy が実際に `repeat()` で長さ 1 の軸を広げることはありません。広げるとメモリを無駄に使うからです。

NumPy には `broadcast_to()` と `broadcast_arrays()` があり、配列をブロードキャストした形で扱えます。`broadcast_to()` は配列を指定した形状に広げ、`broadcast_arrays()` は複数の配列をブロードキャストした結果を返します。どちらも元の配列とデータ領域を共有します。

````python
a = np.arange(0, 60, 10).reshape(-1, 1)
b = np.arange(0, 5)
a2, b2 = np.broadcast_arrays(a, b)   # ブロードキャストした結果
b3 = np.broadcast_to(b, (2, 3, 5))   # 形を (2, 3, 5) に広げる
%C a2.shape; b2.shape; b3.shape
````

ブロードキャストはよく使うので、NumPy には `ogrid` オブジェクトがあり、ブロードキャスト用の配列を作れます。

````python
x, y = np.ogrid[:5, :5]
%C x; y
````

`mgrid` オブジェクトも同じように使えますが、こちらはブロードキャストしたあとの配列を返します。

````python
x, y = np.mgrid[:5, :5]
%C x; y
````

`ogrid` は多次元配列のようにスライスのタプルを添字に使い、ブロードキャストに使える配列の組を返します。スライスには 2 つの書き方があります。

* `開始値:終了値:ステップ`：`np.arange(開始値, 終了値, ステップ)` に似ている
* `開始値:終了値:長さj`：3 番目のパラメータが虚数のときは、返る配列の長さを表す。`np.linspace(開始値, 終了値, 長さ)` に似ている

````python
x, y = np.ogrid[:1:4j, :1:3j]
%C x; y
````

`ogrid` を使うと、等間隔の格子上での 2 変数関数の値を簡単に計算できます。次のプログラムは、3 次元の曲面 $f(x,y) = x e^{-x^2-y^2}$ を計算して描きます。

````python
x, y = np.ogrid[-2:2:20j, -2:2:20j]
z = x * np.exp(-(x**2) - y**2)
````

````python
from matplotlib import pyplot as plt

fig = plt.figure(figsize=(8, 5))
ax = fig.add_subplot(projection="3d")
surf = ax.plot_surface(x, y, z, rstride=1, cstride=1, cmap="coolwarm", linewidth=0.2)
plt.show()
````

ufunc のブロードキャストを活かすには、配列の形を整えることがよくあります。そのために、配列は特別な添字 `None` を受け付けます。`None` の位置に長さ 1 の新しい軸ができます。1 次元配列 `a` なら、`a[None, :]` は `a.reshape(1, -1)`、`a[:, None]` は `a.reshape(-1, 1)` と同じです。

````python
a = np.arange(4)
%C a[None, :]; a[:, None]
````

次の例では、`None` を添字に使ってブロードキャストしています。

````python
x = np.array([0, 1, 4, 10])
y = np.array([2, 3, 8])
x[None, :] + y[:, None]
````

`ix_()` を使うと、2 つの 1 次元配列を、ブロードキャストできる 2 次元配列に変換できます。

````python
gy, gx = np.ix_(y, x)
%C gx; gy; gx + gy
````

この例では、`ix_()` が配列 `x` と `y` をブロードキャストできる 2 次元配列に変換しています。配列 `y` が結果の第 0 軸、配列 `x` が第 1 軸に対応します。`ix_()` の引数は N 個の 1 次元配列で、それらを N 次元空間でブロードキャストできる N 次元配列に変換します。

## ufuncのメソッド

ufunc のオブジェクト自身にもメソッドがあります。これらは入力を 2 つ、出力を 1 つ持つ ufunc でのみ使えます。ほかの ufunc で呼ぶと `ValueError` になります。

`reduce()` の呼び方は `<op>.reduce(array, axis=0, dtype=None)` です。`axis` で指定した軸に沿って、`<op>` をすべての要素の間に挟んだのと同じ計算をします。

````python
r1 = np.add.reduce([1, 2, 3])                       # 1 + 2 + 3
r2 = np.add.reduce([[1, 2, 3], [4, 5, 6]], axis=1)  # (1+2+3), (4+5+6)
%C r1; r2
````

`accumulate()` は `reduce()` に似ていますが、返る配列は入力と同じ形状で、途中の計算結果をすべて保存します。

````python
a1 = np.add.accumulate([1, 2, 3])
a2 = np.add.accumulate([[1, 2, 3], [4, 5, 6]], axis=1)
%C a1; a2
````

`outer()` は、2 つの配列のすべての要素の組について計算します。次のプログラムと同じです。

```python
a.shape += (1,)*b.ndim
<op>(a, b)
a = a.squeeze()
```

`squeeze()` は、配列から長さ 1 の軸を取り除きます。例を見てみます。

````python
np.multiply.outer([1, 2, 3, 4, 5], [2, 3, 4])   # すべての組の積(乗算表)
````

`outer()` の結果は乗算表です。同じ計算を上のプログラムで 1 行ずつ確かめると、乗算表がブロードキャストで計算されていることが分かります。

## まとめ

* ufunc は配列の要素ごとに計算する関数で、C 言語で実装されているため速い。`out` パラメータに配列を渡すと、結果をその配列に書き込める。
* 四則演算・比較演算・論理演算・ビット演算には、それぞれ対応する ufunc がある。ブール配列に `and`、`or`、`not` は使えない。
* ufunc は形の違う配列どうしも計算できる。これをブロードキャストと呼び、長さ 1 の軸を出力の長さに合わせて広げる。
* `None` を添字にすると長さ 1 の軸を追加でき、`ogrid` と `mgrid` で格子状の配列を作れる。
* `reduce()`、`accumulate()`、`outer()` は ufunc のメソッドで、軸に沿った集計や要素の組の計算を行う。
