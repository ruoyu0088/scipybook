### `pyplot` で描く

`pyplot` は MATLAB に似た関数をそろえ、2 次元のグラフを手早く描けます。まず簡単な例です。

:::{admonition} `pylab` モジュール
:class: note
`pylab` は、NumPy と `pyplot` のよく使う関数をまとめたモジュールです。計算と描画を続けて書けるので、対話環境に向いています。この本では `import pylab as pl` で読み込みます。
:::

````python
import matplotlib.pyplot as plt  #❶

x = np.linspace(0, 10, 1000)
y = np.sin(x)
z = np.cos(x**2)

plt.figure(figsize=(8, 4))  #❷

plt.plot(x, y, label="$sin(x)$", color="red", linewidth=2)  #❸
plt.plot(x, z, "b--", label="$cos(x^2)$")  #❹

plt.xlabel("Time(s)")  #❺
plt.ylabel("Volt")
plt.title("PyPlot First Example")
plt.ylim(-1.2, 1.2)
plt.legend()

plt.show()  #❻
````

このプログラムの流れです。❶ `pyplot` を `plt` という名前で読み込みます。❷ `figure()` で `Figure`(図)を作り、それを今の図にします。`figure()` を呼ばずに `plot()` しても、matplotlib が図を作ります。`figsize` は図の幅と高さで、単位はインチです。`dpi` は 1 インチあたりのピクセル数で、既定は 80 なので、この図の幅は `8*80 = 640` ピクセルになります。

❸ 図を作ったら `plot()` で描きます。実際に描く先は `Axes`(サブプロット)です。図に `Axes` がなければ、図のほとんどを覆う `Axes` が作られ、それが今の `Axes` になります。`plot()` の最初の 2 つの引数が x と y のデータです。キーワード引数で線の属性を決めます。

* `label`：凡例に出す名前。前後を `'$'` で囲むと、LaTeX で数式として表示する。
* `color`：色。名前か、`'#'` で始まる 6 桁の 16 進数。`(1.0, 0.0, 0.0)` のような 0〜1 の組でもよい。
* `linewidth`：線の太さ。`lw` とも書ける。

:::{warning}
LaTeX の数式を使うと、描画がかなり遅くなります。
:::

❹ 3 番目の `'b--'` で、色と線の種類をまとめて指定します。`'b'` が青、`'--'` が破線です。対話環境で `plt.plot?` とすると、書式文字列の説明が出ます。

❺ そのあと、`Axes` の属性を関数で設定します。

* `xlabel`、`ylabel`：軸の名前。
* `title`：サブプロットの題。
* `xlim`、`ylim`：軸の表示範囲。
* `legend`：凡例。各線の `label` と線種を並べた箱。

❻ 最後に `plt.show()` で窓を出します。Notebook では要りません。ふつうの環境では、`show()` は窓を閉じるまで止まります。

`plt.savefig()` で、今の図を画像として保存できます。形式は拡張子で決まります。次のプログラムは `test.png` に保存し、`dpi=120` なので、幅は `8*120 = 960` ピクセルになります。

````python
plt.savefig("test.png", dpi=120)
````

:::{tip}
窓を閉じたあとは `savefig()` で保存できません。`show()` を呼ばずに `savefig()` だけ呼んでもかまいません。そうすれば、図をまとめて出力するプログラムが簡単に書けます。
:::

`savefig()` の第 1 引数は、ファイル名でも、ファイルのようなオブジェクトでもかまいません。`io.BytesIO` に書けば、画像の中身を文字列として取り出せます。そのときは `format` で形式を指定します。

````python
import io

buf = io.BytesIO()  # 画像内容を保存するためのBytesIOオブジェクトを作成
plt.savefig(buf, format="png")  # 画像をpng形式でbufに保存
buf.getvalue()[:20]  # 画像内容の最初の20バイトを表示
````

### オブジェクト指向で描く

matplotlib は本来オブジェクト指向のライブラリで、線、文字、目盛りなどの部品は、それぞれオブジェクトとしてメモリにあります。`pyplot` は、その構造を隠して MATLAB 風の関数を提供しています。手早く描くには便利ですが、大きなプログラムには向きません。この章では、オブジェクトを直接扱う書き方を主に説明します。

関数だけの API にするために、`pyplot` の内部には「今の図」「今のサブプロット」が保存されています。`gcf()` で今の `Figure`、`gca()` で今の `Axes` を取れます。Get Current Figure、Get Current Axes の頭文字です。

````python
fig = plt.gcf()
axes = plt.gca()
print(fig, axes)
````

`pyplot` の関数の多くは、今の `Figure` か `Axes` を操作します。`plot()`、`xlabel()`、`savefig()` などです。対話環境で関数名に `??` を付けるとソースが読めます。次は `plot()` のソースで、`gca()` で今の `Axes` を取り、その `plot()` を呼んでいることが分かります。

```python
def plot(*args, **kwargs):
    ax = gca()
    ...
    try:
        ret = ax.plot(*args, **kwargs)
        ...
    finally:
        ax.hold(washold)
```

### 属性を設定する

グラフの部品は、どれもオブジェクトで、属性は `set_*()` か `setp()` で変えられます。`plot()` は `Line2D` のリストを返します。次の例では、その属性を変えます。

````python
fig = plt.figure(figsize=(4, 3))
x = np.arange(0, 5, 0.1)
line = plt.plot(x, 0.05 * x * x)[0]  # plotはリストを返す
line.set_alpha(0.5)  # Line2Dオブジェクトのset_*()メソッドを呼び出して属性値を設定
````

`set_alpha()` で、線を半透明にしました。次のコードは正弦波と余弦波を一度に描き、`lines` に 2 つの `Line2D` が入ります。

````python
lines = plt.plot(x, np.sin(x), x, np.cos(x))
fig
````

`setp()` なら、複数のオブジェクトの属性をまとめて変えられます。

````python
plt.setp(lines, color="r", linewidth=4.0)
fig
````

`get_*()` か `plt.getp()` で、属性を読めます。

````python
print(line.get_linewidth())
print(plt.getp(lines[0], "color"))  # color属性を返す
````

`setp()` と違い、`getp()` は 1 つのオブジェクトだけを扱います。使い方は 2 通りです。

* 属性名を渡す：その属性の値を返す。
* 属性名を渡さない：すべての属性と値を出す。

次の例では、`Figure` の属性を見ます。

````python
plt.getp(fig)
````

`Figure` の `axes` は、図の中のサブプロットのリストです。次のコードで、`gca()` が返す `Axes` と同じものか確かめます。

````python
print(plt.getp(fig, "axes"), plt.getp(fig, "axes")[0] is plt.gca())
````

`getp()` で `Axes` の属性も見られます。`lines` は、そのサブプロットの `Line2D` のリストです。

````python
alllines = plt.getp(plt.gca(), "lines")
print(alllines, alllines[0] is line)  # 最初に描画した曲線が最初の要素
````

この方法で、オブジェクトの属性と、オブジェクトどうしの関係を確かめられます。属性は直接読んでもかまいません。

````python
print(fig.axes, len(fig.axes[0].lines))
````

### 複数のサブプロットを描く

`Figure` には複数のサブプロット(`Axes`)を置けます。`Axes` が描画の領域で、この本ではサブプロットと呼びます。ここまでは 1 つだけでした。`subplot()` で、複数のサブプロットを作れます。

    subplot(numRows, numCols, plotNum)

描画の領域を `numRows` 行 `numCols` 列に等分し、左上から右へ、上から下へ番号を振ります。左上が 1 です。`plotNum` が、作る `Axes` の場所です。3 つとも 10 未満なら、`subplot(323)` のように 1 つの数で書けます。`subplot(3, 2, 3)` と同じ意味です。新しいサブプロットが前のものと重なると、前のものは消えます。

次のプログラムは、3 行 2 列の 6 つのサブプロットを作り、`facecolor` で色を変えます。

````python
fig = plt.figure()
for idx, color in enumerate("rgbyck"):
    plt.subplot(321 + idx, facecolor=color)
````

行や列をまたいで広げたいときは、次のように書きます。

````python
plt.figure()
plt.subplot(221)  # 1行目の左図
plt.subplot(222)  # 1行目の右図
plt.subplot(212)  # 2行目全体
````

窓のツールバーの「Configure Subplots」で、サブプロットの間隔を調整できます。プログラムからは `subplots_adjust()` で同じことができます。引数は `left`、`right`、`bottom`、`top`、`wspace`、`hspace` の 6 つで、値は 0〜1 の、図の幅と高さに対する割合です。

`subplot()` は作った `Axes` を返します。それを変数に入れておき、`sca()` で今の `Axes` に切り替えてから描けます。図を複数使うときは、`figure()` に番号を渡します。同じ番号の図があれば、新しく作らずにそれを今の図にします。次のプログラムは、2 つの図のサブプロットに順に曲線を描きます。

````python
plt.close("all")
````

````python
plt.figure(1, figsize=(6, 3))  # グラフ1を作成
plt.figure(2, figsize=(6, 3))  # グラフ2を作成
ax1 = plt.subplot(121)  # グラフ2にサブプロット1を作成
ax2 = plt.subplot(122)  # グラフ2にサブプロット2を作成

x = np.linspace(0, 3, 100)
for i in range(5):
    plt.figure(1)  #❶ グラフ1を選択
    plt.plot(x, np.exp(i * x / 3))
    plt.sca(ax1)  #❷ グラフ2のサブプロット1を選択
    plt.plot(x, np.sin(i * x))
    plt.sca(ax2)  # グラフ2のサブプロット2を選択
    plt.plot(x, np.cos(i * x))
````

:::{tip}
`sca()` を呼ばずに、`ax1.plot()` のように直接メソッドを呼んでもかまいません。
:::

まず `figure()` で 2 つの図を作り、番号 1 と 2 を付けます。次に図 2 に左右 2 つのサブプロットを作り、`ax1` と `ax2` に入れます。

ループの中では、❶ `figure(1)` で図 1 を今の図にして描きます。❷ `sca(ax1)` と `sca(ax2)` で、それぞれを今のサブプロットにして描きます。サブプロットを選ぶと、それを含む図も今の図になるので、`figure(2)` は要りません。

`subplots()` なら、複数のサブプロットを一度に作れます。次の例の `axes` は形 `(2, 3)` の配列で、要素がサブプロットです。

````python
fig, axes = plt.subplots(2, 3)
[a, b, c], [d, e, f] = axes
print(axes.shape)
print(b)
plt.close(fig)
````

`subplot2grid()` を使うと、表のような複雑な並べ方もできます。Excel の表を作るのに似ています。

```python
subplot2grid(shape, loc, rowspan=1, colspan=1, **kwargs)
```

`shape` が (行, 列)、`loc` がサブプロットの左上の (行, 列)、`rowspan` と `colspan` が占める行数と列数です。次の例は 3×3 の格子に 5 つのサブプロットを作り、それぞれの中央に名前を書きます。

````python
fig = plt.figure(figsize=(6, 6))
ax1 = plt.subplot2grid((3, 3), (0, 0), colspan=2)
ax2 = plt.subplot2grid((3, 3), (0, 2), rowspan=2)
ax3 = plt.subplot2grid((3, 3), (1, 0), rowspan=2)
ax4 = plt.subplot2grid((3, 3), (2, 1), colspan=2)
ax5 = plt.subplot2grid((3, 3), (1, 1))

for idx, ax in enumerate(fig.axes, 1):
    ax.text(0.5, 0.5, "ax{}".format(idx), ha="center", va="center", fontsize=16)
````

### 図に日本語を出す

matplotlib の既定のフォント(DejaVu Sans)には、日本語の字形がありません。そのままでは豆腐(□)になります。設定の方法は 3 つあります。

1. テキストごとにフォントを指定する。
2. プログラムの冒頭で `rcParams` を変える。
3. 設定ファイルを変える。

フォントは名前で指定します。使えるフォントの一覧は、次のコードで取れます。

````python
from matplotlib.font_manager import fontManager

fontManager.ttflist[:6]
````

`ttflist` が matplotlib が見つけたフォントの一覧で、各要素が `Font` です。次のコードで、最初のフォントの名前とファイルを出します。

````python
print(fontManager.ttflist[0].name)
print(fontManager.ttflist[0].fname)
````

次のプログラムは、一覧のフォントで日本語を描きます。

````python
plt.close("all")
````

````python
# %fig=システム内の日本語フォント名を表示
import warnings
from matplotlib.ft2font import FT2Font
from os import path

def has_japanese(fname):
    """かなの字形があるかどうかで、日本語フォントか判断する。"""
    try:
        face = FT2Font(fname)
    except Exception:
        return False
    return all(face.get_char_index(ord(ch)) for ch in "日本語あ")

fonts = []
for font in fontManager.ttflist:
    if not path.exists(font.fname) or font.name in fonts:  #❶
        continue
    if has_japanese(font.fname):  #❷
        fonts.append(font.name)
fonts = sorted(fonts)[:16]  #❸ 多すぎるので先頭 16 個だけ

fig = plt.figure(figsize=(8, 0.5 + 0.45 * len(fonts)))
ax = fig.add_subplot(111)
plt.subplots_adjust(0, 0, 1, 1, 0, 0)
plt.xticks([])
plt.yticks([])
with warnings.catch_warnings():
    warnings.simplefilter("ignore")  # ウェイトが無いフォントの警告を消す
    for i, name in enumerate(fonts[::-1]):
        y = 0.04 + i * 0.045
        ax.text(0.02, y, "日本語フォント", {"fontname": name, "fontsize": 13}, transform=ax.transAxes)
        ax.text(0.55, y, name, {"fontsize": 9}, transform=ax.transAxes)
plt.show()
print(len(fonts), "fonts")
````

❶ 同じフォント名が何度も出るので、名前で重複を除きます。❷ かなの字形があるかどうかを `FT2Font.get_char_index()` で調べ、日本語のフォントだけを残します。ファイルの大きさで絞ると、日本語でない大きなフォントも入ってしまいます。❸ 数が多いので、並べ替えて先頭の 16 個だけを出します。

ここで出てきた名前は、そのまま `fontname` に指定できます。

````python
from matplotlib.font_manager import FontProperties

font = FontProperties(fname=r"c:\windows\fonts\meiryo.ttc", size=14)  #❶
t = np.linspace(0, 10, 1000)
y = np.sin(t)
plt.close("all")
plt.plot(t, y)
plt.xlabel("時間", fontproperties=font)  #❷
plt.ylabel("振幅", fontproperties=font)
plt.title("正弦波", fontproperties=font)
plt.show()
````

❶ `FontProperties` を作り、`fname` にフォントファイルの絶対パスを入れます。❷ `fontproperties` にそのオブジェクトを渡します。

TTC を TTF に分けて、システムのフォントフォルダに入れる方法もあります。matplotlib は起動時にフォントを全部読み直さないので、ファイルを足したら、設定フォルダの `fontlist-*.json` を消してください。

`rcParams` を変えれば、テキストごとに指定しなくても済みます。この本では、環境に合ったフォントを選ぶ `helper.fonts` を使います。

````python
import helper.fonts

plt.rcParams["font.family"] = helper.fonts.current()  # この環境で使える日本語フォント
plt.figure()
plt.plot([1, 2, 3])
plt.xlabel("日本語");
````

## `Artist` オブジェクト

matplotlib は 3 つの層に分かれています。

* `backend_bases.FigureCanvas`：描くための画布。
* `backend_bases.Renderer`：画布に描く方法を知っている。
* `artist.Artist`：Renderer を使って画布に描く方法を知っている。

`FigureCanvas` と `Renderer` は、GUI の部品に描くといった低い層の仕事をします。`Artist` は、図、文字、曲線などの部品の描画と配置を受け持ちます。ふつうは `Artist` だけを触れば済みます。

`Artist` には、単独の部品(`Line2D`、`Rectangle`、`Text`、`AxesImage`)と、ほかの `Artist` をまとめる入れ物(`Axis`、`Axes`、`Figure`)があります。

`Artist` を作って描く手順は、次のとおりです。

* `Figure` を作る。
* その `Figure` に 1 つ以上の `Axes` を作る。
* `Axes` のメソッドで、部品を作る。

`Axis` には、目盛り線、目盛りの文字、グリッド、軸の名前が入ります。目盛りには主目盛りと副目盛りがあり、`get_major_ticks()` と `get_minor_ticks()` で取れます。目盛りは `XTick` か `YTick` で、線と文字を持ちます。`get_ticklabels()` と `get_ticklines()` で、まとめて取れます。

````python
for label in axis.get_ticklabels():
    label.set_color("red")
    label.set_rotation(45)
    label.set_fontsize(16)

for line in axis.get_ticklines():
    line.set_color("green")
    line.set_markersize(25)
    line.set_markeredgewidth(3)
fig
````

副目盛りのリストが空だったのは、位置を決めるものが既定で `NullLocator` だからです。主目盛りは `AutoLocator` で、拡大の状態などから位置を決めます。

````python
print(axis.get_minor_locator())  # 副目盛りの位置を計算するオブジェクト
print(axis.get_major_locator())  # 主目盛りの位置を計算するオブジェクト
````

matplotlib には、目盛りの位置を決める `Locator` と、文字を決める `Formatter` があります。次のプログラムは、x 軸の主目盛りを $\pi/4$、副目盛りを $\pi/20$ にし、主目盛りの文字を $\pi$ で表します。

````python
from fractions import Fraction
from matplotlib.ticker import MultipleLocator, FuncFormatter  #❶

x = np.arange(0, 4 * np.pi, 0.01)
fig, ax = plt.subplots(figsize=(8, 4))
plt.plot(x, np.sin(x), x, np.cos(x))


def pi_formatter(x, pos):  #❷
    frac = Fraction(int(np.round(x / (np.pi / 4))), 4)
    d, n = frac.denominator, frac.numerator
    if frac == 0:
        return "0"
    elif frac == 1:
        return r"$\pi$"
    elif d == 1:
        return r"${%d} \pi$" % n
    elif n == 1:
        return r"$\frac{\pi}{%d}$" % d
    return r"$\frac{%d \pi}{%d}$" % (n, d)


# 2つの軸の範囲を設定
plt.ylim(-1.5, 1.5)
plt.xlim(0, np.max(x))

# 図の下部の余白を設定
plt.subplots_adjust(bottom=0.15)

plt.grid()  # グリッドを有効にする

# 主目盛りをpi/4に設定
ax.xaxis.set_major_locator(MultipleLocator(np.pi / 4))  #❸

# 主目盛りテキストをpi_formatter関数で計算
ax.xaxis.set_major_formatter(FuncFormatter(pi_formatter))  #❹

# 副目盛りをpi/20に設定
ax.xaxis.set_minor_locator(MultipleLocator(np.pi / 20))  #❺

# 目盛りテキストのサイズを設定
for tick in ax.xaxis.get_major_ticks():
    tick.label1.set_fontsize(16)
````

❶ 目盛りの位置と文字のクラスは `matplotlib.ticker` にあります。ここでは 2 つ読み込みます。

* `MultipleLocator`：❸❺ 指定した値の整数倍に、目盛りを置く。
* `FuncFormatter`：❹ 指定した関数で文字を作る。関数には目盛りの値と番号が渡る。❷ `pi_formatter()` がその関数で、値に応じて $\pi$ の分数を返す。
