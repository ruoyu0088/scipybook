# Panel で UI を作る

Panel は、ブラウザで動くダッシュボードの UI を作るためのライブラリです。土台は Bokeh の部品とレイアウトで、Notebook や Bokeh サーバーの中で、画面と Python のあいだのやり取りをします。

まず HoloViews と Panel を読み込み、`pn.extension()` で Notebook 用のプラグインを初期化します。引数は読み込むプラグインの名前で、ここでは数式の katex と、コード表示の ace を読み込みます。

- `interact`、`bind`、`param` で UI を作る 4 つの書き方
- フーリエ級数の UI
- Widget / Pane / Panel の 3 つの部品
- Param でパラメータを定義する

```python
import panel as pn
import numpy as np
import holoviews as hv

hv.extension("bokeh")
pn.extension("katex")
```

## UI の作り方

次の `plot_sin_wave()` は、4 つの引数から減衰する正弦波を作り、HoloViews の曲線を返します。この関数の 4 つの引数を、部品で変えられるようにします。

```python
def plot_sin_wave(freq, amp, phase, dump):
    t = np.linspace(0, 10, 400)
    y = amp * np.exp(-dump * t) * np.sin(2 * np.pi * freq * t + np.deg2rad(phase))
    return hv.Curve((t, y)).opts(width=600, show_grid=True)
```

### `interact`

`interact` は、関数の引数を動かせる部品を自動で作ります。キーワードで、各引数の範囲を決めます。

渡す値によって、部品の種類が変わります。

* タプル `(min, max)`：連続の値を動かすスライダー。
* `range` やリスト：決まった値から選ぶスライダー。

```python
ui = pn.interact(
    plot_sin_wave, freq=(1, 10), amp=(1, 10), phase=range(0, 360, 45), dump=(0.0, 1.0)
)
ui
```

できた UI の中身を見ます。

* 外側の `Column` が全体の入れ物。
* 中の `Column` が部品を並べる入れ物。
* 中の `Row` に、グラフの HoloViews オブジェクトが入る。

```python
print(ui)
```

`Column` や `Row` は、`objects` に中身を入れ子で持ちます。次のコードでは、外側を `Row` にして、部品とグラフを左右に並べます。中身は同じオブジェクトなので、新しい UI で値を変えると、元のグラフも変わります。

```python
pn.Row(*ui.objects)
```

### コールバックで作る

`interact()` は手軽ですが、細かいことはできません。込み入った UI は、部品とグラフを自分で作り、つなぎます。つなぎ方で一番よく使うのがコールバックです。

❶ 部品の `value` が変わったときに呼ぶ関数を登録します。❷ その関数の中で `plot_sin_wave()` を呼び、結果を `fig_panel.object` に入れて図を差し替えます。

```python
freq = pn.widgets.FloatSlider(value=5, start=1, end=10, name="freq")
amp = pn.widgets.FloatSlider(value=1, start=1, end=5, name="amp")
phase = pn.widgets.Select(
    value=90, options=np.arange(0, 360, 45).tolist(), name="phase"
)
dump = pn.widgets.Spinner(value=0.5, step=0.1, name="dump")

ctrl_panel = pn.Column(freq, amp, phase, dump)
fig_panel = pn.pane.HoloViews()


def update_figure(*events):
    fig_panel.object = plot_sin_wave(
        freq.value, amp.value, phase.value, dump.value
    )  # ❷


for widget in [freq, amp, phase, dump]:
    widget.param.watch(update_figure, "value")  # ❶

update_figure()  # ❸
pn.Row(ctrl_panel, fig_panel)
```

### `bind()` で作る

コールバックのコードは込み入りがちです。`bind()` を使うと、関数と部品を結び付けられ、短く書けます。

❶ `bind()` で `plot_sin_wave` を 4 つの部品と結びます。部品の値が変わるたびに、関数が呼ばれます。❷ 戻り値は `binded_func` という特別な関数で、HoloViews の pane にそのまま渡せます。

```python
freq = pn.widgets.FloatSlider(value=5, start=1, end=10, name="freq")
amp = pn.widgets.FloatSlider(value=1, start=1, end=5, name="amp")
phase = pn.widgets.Select(
    value=90, options=np.arange(0, 360, 45).tolist(), name="phase"
)
dump = pn.widgets.Spinner(value=0.5, step=0.1, name="dump")

ctrl_panel = pn.Column(freq, amp, phase, dump)
binded_func = pn.bind(plot_sin_wave, freq, amp, phase, dump)  # ❶
fig_panel = pn.pane.HoloViews(binded_func)  # ❷
pn.Row(ctrl_panel, fig_panel)
```

### `Parameterized` で作る

ここまでは部品を手で作りました。`param.Parameterized` を継承すると、パラメータを 1 つのクラスにまとめ、その定義から UI を自動で作れます。Param の詳しい話はこの章の最後にあります。

次のコードは `WaveParam` を作り、`param.Number` や `param.Selector` で属性を決めます。❶ `WaveParam` のオブジェクトを作ります。❷ これを `Row()` に渡すと、部品が自動でできます。❸ `bind()` には `pars.param` を渡します。

```python
import param


class WaveParam(param.Parameterized):
    freq = param.Number(default=5, bounds=(1, 10))
    amp = param.Number(default=1, bounds=(1, 5))
    phase = param.Selector(default=90, objects=np.arange(0, 360, 45).tolist())
    dump = param.Number(default=0.5, step=0.1)


pars = WaveParam()  # ❶
layout = pn.Row(
    pars,  # ❷
    pn.bind(
        plot_sin_wave, *[pars.param[name] for name in ["freq", "amp", "phase", "dump"]]
    ),  # ❸
)
layout
```

できた UI の中身は、次のとおりです。

```python
print(layout)
```

グラフを更新する関数を、クラスのメソッドにすることもできます。❶ `bind()` の代わりに `depends()` でメソッドを飾ると、属性と結び付きます。❷ `pars` と `pars.plot` を `Row` に渡せば、UI とグラフが一緒にできます。

```python
class WaveParam(param.Parameterized):
    freq = param.Number(default=5, bounds=(1, 10))
    amp = param.Number(default=1, bounds=(1, 5))
    phase = param.Selector(default=90, objects=np.arange(0, 360, 45).tolist())
    dump = param.Number(default=0.5, step=0.1)

    @pn.depends("freq", "amp", "phase", "dump")  # ❶
    def plot(self):
        return plot_sin_wave(self.freq, self.amp, self.phase, self.dump)


pars = WaveParam()
layout = pn.Row(pars, pars.plot)  # ❷
layout
```

## フーリエ級数の UI

この例では、いろいろな関数のフーリエ級数の波形を UI で見ます。`function` というドロップダウンで関数を、`order` というスライダーで次数を選びます。展開した式が下に、曲線が右に出ます。

計算の部分です。`calc_fourier(func, order)` は、関数 `func` をフーリエ級数に展開し、最初の `order` 項を返します。戻り値は 2 つで、LaTeX の式と、その項の `Curve` です。何度も呼ぶので、`functools.lru_cache()` で結果を覚えます。

```python
from functools import lru_cache
import numpy as np
from sympy import Piecewise, S, Symbol, And, fourier_series, lambdify, latex, pi


@lru_cache()
def calc_fourier(func, order):
    x = Symbol("x")
    fs = fourier_series(func, (x, 0, 2))
    fs_t = fs.truncate(n=order)
    fs_lambda = lambdify([x], fs_t)
    x_ = np.linspace(0, 4, 200)
    y_ = fs_lambda(x_)
    return f"$${latex(fs_t)}$$", hv.Curve((x_, y_))


x = Symbol("x")

functions = {
    "step": Piecewise((1, x < 1), (0, True)),
    "triangle": Piecewise((x, x < 1), (2 - x, True)),
    "pulse": Piecewise((1, And(S(2) / 3 < x, x < S(4) / 3)), (0, True)),
    "trapezoid": Piecewise(
        (S(3) / 2 * x, x < S(2) / 3), (1, x < S(4) / 3), (3 - S(3) / 2 * x, True)
    ),
    "sawtooth": x,
}
```

階段関数のフーリエ級数の、最初の 5 項の式とグラフです。

```python
from IPython.display import display_latex

formula, curve = calc_fourier(functions["step"], 5)
display_latex(formula, raw=True)
curve.opts(width=600, height=300)
```

### リアクティブに作る

`calc_fourier()` の UI を作ります。部品はまとめて `FourierSeriesViewer` クラスに入れます。

❶ 2 つの部品、ドロップダウンと整数のスライダーを作ります。❷ `bind()` で `calc_fourier` と 2 つの部品を結びます。❸ 戻り値は 2 つなので、そのまま pane には渡せません。`rx()` のオブジェクトに `[]` を付けて、1 つ目を `LaTeX`、2 つ目を `HoloViews` に渡します。❹ 最後に、UI のレイアウトを組みます。

```python
class FourierSeriesViewer:
    def __init__(self, functions):
        self.functions = functions
        self.func_name = pn.widgets.Select(name="function", options=self.functions)  # ❶
        self.order = pn.widgets.IntSlider(
            name="order", value=5, start=1, end=10, step=1
        )
        func = pn.bind(calc_fourier, self.func_name, self.order)  # ❷
        self.latex_pane = pn.pane.LaTeX(func.rx()[0])  # ❸
        self.curve_pane = pn.pane.HoloViews(func.rx()[1].opts(width=600, height=300))
        self.layout = pn.Column(  # ❹
            pn.Row(pn.Column(self.func_name, self.order), self.curve_pane),
            self.latex_pane,
        )


fsv = FourierSeriesViewer(functions)
fsv.layout
```

`rx()` は、関数や式を「リアクティブ」にします。リアクティブなら、入力が変わると出力を計算し直します。`bind()` は引数と戻り値だけですが、`rx()` は Python の計算全体をリアクティブにできます。

次の `func()` は 2 つの値を返します。`pn.bind(func, x=slider).rx()[1]` で、2 つ目の戻り値をスライダーと結びます。

```python
slider = pn.widgets.FloatSlider(name="x", value=5, start=1, end=10)


def func(x):
    return [x + 1, x * 10]


pn.bind(func, x=slider).rx()[1]
```

### コールバックで作る(パラメータ版)

`param.Parameterized` の派生クラスに `Parameter` を置けば、部品の作成とイベントの処理が簡単になります。次の `FourierSeriesViewer2` は、`func` と `order` の 2 つの `Parameter` を持ちます。

`__init__()` では、❶ `self.param.func.objects` で候補の値を入れます。❷ UI を出す前に `latex_pane` と `curve_pane` の中身を決める必要があるので、`func` に最初の値を入れた時点で `calc()` が走ります。

```python
import param


class FourierSeriesViewer2(param.Parameterized):
    func = param.Selector()
    order = param.Integer(default=5, bounds=(1, 10))

    def __init__(self, functions, **kw):
        super().__init__(**kw)
        self.param.func.objects = functions  # ❶

        self.latex_pane = pn.pane.LaTeX()
        self.curve_pane = pn.pane.HoloViews()
        self.layout = pn.Column(pn.Row(self, self.curve_pane), self.latex_pane)

        self.func = next(iter(functions.values()))  # ❷

    @param.depends("func", "order", watch=True)  # ❸
    def calc(self):
        order = self.order
        formula, curve = calc_fourier(self.func, order)
        self.latex_pane.object = formula
        self.curve_pane.object = curve.opts(width=500, show_grid=True)


fsv2 = FourierSeriesViewer2(functions)
fsv2.layout.servable()
```

## Param でパラメータを定義する

Panel のクラスは `param.Parameterized` から派生していて、属性はすべて Parameter です。この節では、Param の使い方を説明します。

Python では、`__init__()` で属性を初期化しますが、値を持てるだけです。Param で属性を定義すると、型、値の範囲、説明も持たせられます。

### Parameter 属性

Parameter を持つクラスは `param.Parameterized` から派生し、属性は `param.Parameter` の派生クラスで定義します。

次の例では、❶ `Calculator` は `param.Parameterized` から派生し、`op1`、`op2`、`operator`、`result` の 4 つの属性を持ちます。❷ `op1` と `op2` は `Number` で、`default`、`step`、`bounds`、`label` で既定値、刻み、範囲、表示名を決めます。❸ `operator` は `Selector` で、選べる値は `objects` に並べます。

```python
import param


class Calculator(param.Parameterized):  # ❶
    op1 = param.Number(default=2.0, step=0.1, bounds=(-10, 10), label="演算数1")  # ❷
    op2 = param.Number(default=3.0, step=0.1, bounds=(-10, 10), label="演算数2")
    operator = param.Selector(objects=list("+-*/"), label="演算子")  # ❸
    result = param.String(label="結果")  # ❹
```

Parameter のクラスは、すべて `param.Parameter` から派生しています。よく使うものは次のとおりです。

| 型 | 説明 | 主な引数 |
|:--|:--|:--|
| `String`  | 文字列  | `regex`、`allow_None` |
| `Number`  | 数値  | `bounds`、`step` |
| `Boolean`  | 真偽値  | |
| `Filename`  | ファイルのパス | `search_paths` |
| `List` | リスト | `class_` |
| `Tuple` | タプル | `length` |
| `DataFrame` | pandas の `DataFrame` | |
| `ListSelector` | 一覧から選ぶ | `objects` |
| `Selector` | 1 つ選ぶ | `objects` |
| `Range` | 範囲 | `bounds` |

```python
print_subclasses(param.parameterized.Parameter, as_html=True, height=300)
```

`Calculator` のオブジェクトを作って中身を見ます。属性の値が表示され、`name` も自動で付いています。値はどれも既定値です。

```python
c1 = Calculator()
c1
```

### コールバック

Parameter に値を入れたときに、決めた関数を呼べます。

❶ `c1.param.watch()` で、`c1` の Parameter に関数を登録します。第 1 引数が関数、第 2 引数が属性名か、そのリストです。`onlychanged=False` なら代入のたびに、`True` なら値が変わったときだけ呼びます。戻り値は `Watcher` です。

❷ 渡る引数の数は決まらないので、関数は `*events` で受けます。各引数は `Event` で、`name`、`old`、`new` を持ちます。

```python
def callback(*events):  # ❷
    for event in events:
        print(
            f"obj: {event.obj.name if event.obj is not None else None}, ",
            f"what: {event.what}, name: {event.name}, "
            f"old: {event.old}, new: {event.new}, type: {event.type}",
        )
    print()


c1 = Calculator(op1=0, op2=0, name="c1")
watch = c1.param.watch(callback, ["op1", "op2"], onlychanged=False)  # ❶
print("c1.op1 = 5")
c1.op1 = 5  # ❸

print("c1.update(op1=3, op2=6)")
c1.param.update(op1=3, op2=6)  # ❹

print("batch_watch() for loop")
with param.parameterized.batch_call_watchers(c1):  # ❺
    for i in range(8):
        c1.op1 = i
        c1.op2 = i + 1

c1.param.unwatch(watch)  # ❻
print("c1.op1 = 6")
c1.op1 = 6
```

`watch()` の `what` で、値以外の情報の変化も受けられます。次のプログラムは `op2` の範囲が変わったときに関数を呼びます。

```python
c1.param.watch(callback, "op2", what="bounds")
c1.param.op2.bounds = (0, 1000)
```

クラスを定義するときに、`param.depends()` でメソッドをコールバックにできます。

❶ `Calculator2` は `Calculator` を継承し、`__init__()` で `**kw` を受けます。❷ 属性を正しく入れるために、ほかの初期化より先に親の `__init__()` を呼びます。❸ `self.param.trigger()` で `operator` のイベントを起こします。❹ `param.depends()` で、`run()` を `op1`、`op2`、`operator` に結びます。`watch=True` のときだけ、コールバックとして登録されます。

```python
class Calculator2(Calculator):  # ❶
    def __init__(self, **kw):
        super().__init__(**kw)  # ❷
        self.param.trigger("operator")  # ❸

    @param.depends("op1", "op2", "operator", watch=True)  # ❹
    def run(self):
        import operator as op

        funcs = {"+": op.add, "-": op.sub, "*": op.mul, "/": op.truediv}
        result = funcs[self.operator](self.op1, self.op2)
        self.result = f"{self.op1:4.1f} {self.operator} {self.op2:4.2f} = {result:4.2f}"


c2 = Calculator2()
print(c2.result)  # ❺
c2.op1 = 5
c2.operator = "*"
print(c2.result)
```

### UI を自動で作る

`Parameterized` の派生クラスは、UI を自動で作るのによく使います。まず Panel を読み、`extension()` で Notebook のプラグインを初期化します。

```python
import panel as pn

pn.extension(inline=False)
```

`panel()` で `Calculator2` の `c2` の UI を作ります。`op1` と `op2` は範囲があるのでスライダー、`operator` はドロップダウン、`result` は入力欄になります。`op1`、`op2`、`operator` を変えると、オブジェクトの値が変わり、`run()` が走ります。`run()` が `result` を入れると、入力欄もすぐ変わります。

```python
p = pn.panel(c2, name="計算器")
p
```

## まとめ

* Panel は Bokeh の上に乗る UI のライブラリ。`pn.extension()` で初期化する。
* UI の作り方は 4 通り。`interact()`(自動)、コールバック、`bind()`、`param.Parameterized`(自動)。
* `bind()` は関数と部品を結ぶ。`rx()` を使えば計算の途中もリアクティブにできる。
* 部品は Widget、Pane、Panel の 3 層。`panel.widgets`、`panel.pane`、`panel.layout` にある。
* Param の Parameter は、型・範囲・説明を持つ属性。`watch()` と `depends()` で変化を受けられる。
* `panel(obj)` で、`Parameterized` のオブジェクトから UI を自動で作れる。
