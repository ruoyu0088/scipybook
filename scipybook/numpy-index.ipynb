## 添字のタプル

多次元配列の添字は、配列の次元数と同じ長さのタプルになります。添字のタプルが次元数より長いとエラーになり、短い場合は足りない分に `:` を補って次元数に合わせます。

添字がタプルでない場合、NumPy はそれを 1 つの要素を持つタプルとして扱います。変換と `:` の補完を経て、標準的な添字のタプルができます。タプルの各要素は、スライス、整数、整数配列、ブール配列のいずれかです。これら以外の要素、例えばリストやタプルは、整数配列に変換されます。

例えば 3 次元配列 `a` に、2 次元のリスト `list_index` と、それを `tuple()` でタプルにした `tuple_index` を添字として渡すと、結果が変わります。NumPy はリスト `list_index` を `(list_index, :, :)` と解釈し、`tuple_index` は `(tuple_index[0], tuple_index[1], :)` として扱うためです。

````python
a = np.arange(3 * 4 * 5).reshape(3, 4, 5)
list_index = [[0], [1]]
tuple_index = tuple(list_index)
%C a[list_index]; a[tuple_index]
````

添字のタプルの要素がすべてスライスと整数なら、結果は元の配列のビューになります。つまり、データ領域を元の配列と共有します。

## 整数配列を添字に使う

次に、添字のタプルがスライスと整数配列でできている場合を考えます。整数配列が $N_t$ 個、スライスが $N_s$ 個あり、$N_t+N_s$ が配列の次元数 $D$ だとします。

まず、$N_t$ 個の整数配列がブロードキャストできる必要があります。ブロードキャストしたあとの次元数を $M$、形状を $(d_0, d_1, ..., d_{M-1})$ とします。

$N_s$ が 0、つまりスライスがない場合、結果の配列 `result` の形状は、整数配列をブロードキャストしたあとの形状と同じです。各要素の値は次の式で決まります。

$$result[i_0, i_1, ..., i_{M-1}] = X[ind_0[i_0, i_1, ..., i_{M-1}], ..., ind_{N_t-1}[i_0, i_1, ..., i_{M-1}]]$$

ここで $ind_0$ から $ind_{N_t-1}$ は、ブロードキャストしたあとの整数配列です。例で確かめます。

```{tip}

ある軸に沿って整数配列で要素を取り出すだけなら、`numpy.take()` 関数が使えます。整数配列の添字より少し速く、範囲外の添字の処理にも対応しています。
```

````python
idx0 = np.array([[1, 2, 1], [0, 1, 0]])
idx1 = np.array([[[0]], [[1]]])
idx2 = np.array([[[2, 3, 2]]])
b = a[idx0, idx1, idx2]
b
````

`idx0`、`idx1`、`idx2` の `shape` は、それぞれ `(2, 3)`、`(2, 1, 1)`、`(1, 1, 3)` です。ブロードキャストのルールに従い、長さが 3 に満たない `shape` の前に 1 を補い、次元数をそろえます。ブロードキャストしたあとの `shape` は、各軸の最大値になります。

```
(1, 2, 3)
(2, 1, 1)
(1, 1, 3)
---------
 2  2  3
```

つまり 3 つの整数配列をブロードキャストすると `(2, 2, 3)` になり、これが結果の配列の形状です。

````python
b.shape                     # (2, 2, 3)：idx0〜idx2 をブロードキャストした形
````

`broadcast_arrays()` を使うと、ブロードキャストしたあとの配列を確認できます。

````python
ind0, ind1, ind2 = np.broadcast_arrays(idx0, idx1, idx2)
%C ind0; ind1; ind2
````

結果の任意の要素 `b[i, j, k]` は、配列 `a` を `ind0`、`ind1`、`ind2` で引いた値と同じです。

````python
i, j, k = 0, 1, 2
print(b[i, j, k], a[ind0[i, j, k], ind1[i, j, k], ind2[i, j, k]])

i, j, k = 1, 1, 1
print(b[i, j, k], a[ind0[i, j, k], ind1[i, j, k], ind2[i, j, k]])
````

次に、$N_s$ が 0 でない場合を考えます。スライスがあると話が複雑になるので、2 つの場合に分けます。

1 つ目は、添字のタプルの中で整数配列が続いている場合です。整数配列が 1 つだけ、または連続しているときは、結果の `shape` は、元の配列の `shape` のうち整数配列が占める部分を、ブロードキャストしたあとの `shape` に置き換えたものになります。例えば元の配列 `a` の `shape` が `(3, 4, 5)` で、`idx0` と `idx1` をブロードキャストした形が `(2, 2, 3)` なら、`a[1:3, idx0, idx1]` の形状は `(2, 2, 2, 3)` になります。

````python
c = a[1:3, idx0, idx1]     # 整数配列が連続している場合
c.shape                     # (2, 2, 2, 3)：スライスの 2 と、ブロードキャストした (2, 2, 3)
````

`c` の `shape` の最初の 2 はスライス `1:3` の長さで、後ろの `(2, 2, 3)` は `idx0` と `idx1` をブロードキャストした形です。

````python
ind0, ind1 = np.broadcast_arrays(idx0, idx1)
ind0.shape                  # (2, 2, 3)：idx0 と idx1 をブロードキャストした形
````

````python
i, j, k = 1, 1, 2
print(c[:, i, j, k])
print(a[1:3, ind0[i, j, k], ind1[i, j, k]])   # c[:, i, j, k] と同じ値
````

2 つ目は、整数配列が連続していない場合です。このときは、整数配列をブロードキャストした形に、スライスの軸の長さを後ろに付け加えた形状になります。例えば `a[idx0, :, idx1]` の `shape` は `(2, 2, 3, 4)` です。ここで `(2, 2, 3)` は `idx0` と `idx1` をブロードキャストした形で、4 は配列 `a` の第 1 軸の長さです。

````python
d = a[idx0, :, idx1]       # 整数配列が連続していない場合
d.shape                     # (2, 2, 3, 4)：整数配列の形の後ろに、スライスの軸が付く
````

````python
i, j, k = 1, 1, 2
%C d[i, j, k, :]; a[ind0[i, j, k], :, ind1[i, j, k]]
````

## 応用例：軸ごとに長さの違う範囲を取り出す

ここまでの規則を使って、NumPy のメーリングリストで話題になった問題を解いてみます。

```{seealso}
http://mail.scipy.org/pipermail/numpy-discussion/2008-July/035764.html

NumPy メーリングリストの元の投稿
```

問題を少し簡単にします。形状が `(I, J, K)` の 3 次元配列 `v` と、形状が `(I, J)` の 2 次元配列 `idx` があり、`idx` の値は 0 から `K-L` までの整数です。添字の計算で、次の条件を満たす配列 `r` を作ります。

```python
r[i, j, :] = v[i, j, idx[i, j]:idx[i, j]+L]
```

次の図のように、左の図の塗りつぶした部分を取り出して、右の図のような配列を得ます。

![三次元配列のインデックス演算問題の図](images/numpy_array_index.png)

まず、確認しやすい配列 `v` を作ります。第 2 軸の各層の値がその層の番号になるようにするので、`v[:, :, i]` の要素はすべて `i` です。次に、配列 `idx` を乱数で作ります。各要素は 0 から `K-L` までの整数です。

````python
I, J, K, L = 6, 7, 8, 3
_, _, v = np.mgrid[:I, :J, :K]

rng = np.random.default_rng(0)          # 毎回同じ結果になるように種を固定する
idx = rng.integers(0, K - L, size=(I, J))
````

次に、配列 `idx` から、第 2 軸の添字になる配列 `idx_k` を作ります。形状は `(I, J, L)` で、第 2 軸の各層の値は `idx` に層の番号を足したもの、つまり `idx_k[:, :, i] = idx + i` です。

````python
idx_k = idx.reshape(I, J, 1) + np.arange(L)
idx_k.shape
````

続いて、第 0 軸と第 1 軸の添字になる配列を作ります。`shape` はそれぞれ `(I, 1, 1)` と `(1, J, 1)` です。

````python
idx_i, idx_j, _ = np.ogrid[:I, :J, :K]   # (I,1,1) と (1,J,1) の添字
````

`idx_i`、`idx_j`、`idx_k` を添字にして配列 `v` から要素を取り出すと、求める配列になります。

````python
r = v[idx_i, idx_j, idx_k]

i, j = 2, 3          # 1 組だけ取り出して確認する
%C r[i, j, :]; v[i, j, idx[i, j]:idx[i, j] + L]
````

結果の `r[i, j, :]` が、元の配列 `v` の該当する範囲と一致しています。すべての要素を確かめたいときは、この部分をループに変えてください。

## ブール配列を添字に使う

ブール配列をそのまま添字にする場合、または添字のタプルにブール配列が入っている場合、NumPy は `nonzero()` でブール配列を整数配列の組に変換し、その整数配列で要素を取り出します。

`nonzero(a)` は、配列 `a` で値が 0 でない要素の添字を返します。戻り値は長さが `a.ndim`(配列 `a` の軸数)のタプルで、各要素は整数配列です。値は、0 でない要素の各軸での添字になります。例えば 1 次元のブール配列 `b1` なら、`nonzero(b1)` は長さ 1 のタプルを返し、`b1[0]` と `b1[2]` の値が 0 でないことが分かります。

```{tip}
ある軸に沿ってブール配列で要素を取り出すだけなら、`numpy.compress()` 関数が使えます。
```

````python
b1 = np.array([True, False, True, False])
np.nonzero(b1)              # 0 でない要素の添字。戻り値はタプル
````

2 次元配列 `b2` なら、`nonzero(b2)` は長さ 2 のタプルを返します。第 0 要素が 0 でない要素の第 0 軸の添字、第 1 要素が第 1 軸の添字です。次の結果から、`b2[0, 0]`、`b2[0, 2]`、`b2[1, 0]` の値が 0 でないことが分かります。

````python
b2 = np.array(
    [
        [True, False, True, False],
        [True, False, False, False],
        [False, False, False, False],
    ]
)
np.nonzero(b2)
````

ブール配列をそのまま添字にすることは、`nonzero()` が返すタプルを添字にすることと同じです。

````python
a = np.arange(3 * 4 * 5).reshape(3, 4, 5)
%C a[b2]; a[np.nonzero(b2)]
````

添字のタプルにブール配列が入っている場合は、そのブール配列を `nonzero()` が返す整数配列に展開したのと同じになります。

````python
b3 = np.array(
    [
        [True, False, True, False, False],
        [True, False, False, False, False],
        [False, False, False, False, False],
        [False, False, False, False, False],
    ]
)
%C a[1:3, b3]; a[1:3, np.nonzero(b3)[0], np.nonzero(b3)[1]]
````

## まとめ

* 多次元配列の添字はタプルとして解釈され、足りない軸には `:` が補われる。
* 整数配列を添字にすると、整数配列をブロードキャストした形の配列が返る。スライスと混ざると、整数配列が連続しているかどうかで結果の形状が変わる。
* スライスと整数だけで添字を作るとビューになり、整数配列を使うとコピーになる。
* ブール配列は `nonzero()` で整数配列の組に変換してから添字として使われる。
