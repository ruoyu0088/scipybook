# Bokeh の基本

Bokeh は、ブラウザで図を表示するためのライブラリです。描画はブラウザの JavaScript が行い、Python は図のデータ構造を作り、データを更新します。

matplotlib と比べると、Bokeh は対話的な操作に強いです。ツールチップ、ズーム、移動を簡単に付けられます。HTML ファイルとして保存でき、Flask や Django のアプリにも組み込めます。matplotlib は静止画が中心で、対話的な機能は弱いです。

## 図の表示のしかた

Bokeh の使い方は 4 通りあります。

1. Notebook の中で図を作り、セルの出力に表示する。
2. HTML ファイルに書き出す。図の情報は JSON で HTML に入るので、Python のない PC でもブラウザで見られ、移動やズームもできる。
3. Notebook に表示したあと、Python でデータを変えてブラウザに送り、図を更新する。
4. Bokeh サーバーで、Python のスクリプトをネットワークに公開する。ネットワークのどこからでも URL で開ける。ボタンや入力欄、選択欄などの部品があるので、複数の人が同時に使うアプリも作れる。

この節では、簡単な例でこの 4 つを紹介します。

### Notebook に表示する

Notebook に Bokeh の図を出すには、まず `output_notebook()` を実行します。これで、Bokeh の JavaScript をネットワークから読み込みます。ネットにつながらないときは、この章の最後の節を見てください。

````python
import bokeh
from bokeh.plotting import output_notebook, figure, show, save

output_notebook()
````

次のプログラムは 2 本の曲線を描きます。❶ `figure()` で `Figure` の `fig` を作ります。❷ `line()` で曲線のオブジェクトを作ります。2 つの位置引数が、各点の x と y です。❸ `fig.legend[0]` で凡例を取り、`update()` で属性を決めます。❹ `show()` で表示します。

属性は、作るときのキーワード引数でも指定できます。`x_range`、`color`、`line_width` などです。`update()` なら、`location` や `orientation` のようにまとめて指定できます。

````python
def plot_demo():
    x = np.linspace(-2, 2, 400)
    fig = figure(width=500, height=300, title="正弦余弦曲線", x_range=(0, 1))  # ❶
    line_sin = fig.line(
        x,
        np.sin(2 * np.pi * x),
        color="red",
        legend_label="sin",
        line_width=2,
        name="sin",
    )  # ❷
    line_cos = fig.line(
        x,
        np.cos(2 * np.pi * x),
        color="green",
        legend_label="cos",
        line_width=2,
        name="cos",
    )
    fig.legend[0].update(location="bottom_left", orientation="horizontal")  # ❸
    return fig, (line_sin, line_cos)


fig, _ = plot_demo()
show(fig)  # ❹
````

`save()` で、図を HTML ファイルに保存できます。上のコードを何度も呼ぶと、ブラウザでは新しい図が古い図を上書きしますが、Python の中では `Figure` がいくつもでき、どれも今の `Document` に入ります。`save()` は、その `Document` の図をすべて HTML にします。

❶ 今の `Document` は `curdoc()` で取れ、`clear()` で中身を消せます。❷ `save()` に `resources=CDN` を渡すと、Bokeh の JavaScript と CSS をネットの CDN から読みます。HTML は小さくなりますが、見るときにネットが要ります。

````python
from bokeh.resources import CDN
from bokeh.plotting import curdoc

curdoc().clear()  # ❶
save(fig, "sincos.html", resources=CDN, title="Plots");  # ❷
````

### JavaScript で図を更新する

Bokeh には、画面を作るための部品がそろっていて、JavaScript でイベントに応答できます。次の例では、スライダーを動かすと、2 本の曲線の周波数が変わります。更新は JavaScript が行うので、Python がなくても動きます。

````python
from bokeh.models import Slider, CustomJS, GraphRenderer
from bokeh.layouts import column

fig, (line_sin, line_cos) = plot_demo()
freq_slider = Slider(title="freq", start=1, end=10, step=0.1, value=1)
layout = column(freq_slider, fig)

js_code = """
let freq = cb_obj.value;
let data_sin = line_sin.data_source.data;
let data_cos = line_cos.data_source.data;

data_sin.y = data_sin.x.map(x => Math.sin(2*Math.PI*freq*x));
data_cos.y = data_cos.x.map(x => Math.cos(2*Math.PI*freq*x));

line_sin.data_source.change.emit();
line_cos.data_source.change.emit();
"""

update_plot = CustomJS(args=dict(line_sin=line_sin, line_cos=line_cos), code=js_code)
freq_slider.js_on_change("value", update_plot)
show(layout)
````

イベントの処理はブラウザの JavaScript で動くので、HTML に書き出してもスライダーは働きます。次のプログラムは `layout` を HTML に保存し、ブラウザで開きます。スライダーを動かして、曲線が変わることを確かめてください。

````python
import os
import webbrowser

curdoc().clear()
html_fn = save(layout, "sincos_freq.html", resources=CDN, title="Plots")
if os.environ.get("SCIPYBOOK_OPEN_BROWSER"):  # 手元で見るときだけ開く
    webbrowser.open_new_tab(html_fn)
print(html_fn)
````

### Python で図を更新する

ここまでの方法は一度きりで、Python で作った構造を JSON にしてブラウザに送り、表示します。表示したあとは、Python から図を変えられません。

次の例では、❶ スライダーの `on_change()` で、`value` が変わったときの関数を登録します。❷ その関数には、属性名、元の値、新しい値の 3 つが渡ります。関数の中で `data_source` を取り、`data` の y を書き換えます。

`show(layout)` でスライダーと図は出ますが、動かしても図は変わりません。Python の関数は Bokeh サーバーの中でしか働かないからです。

Notebook で関数を試すには、`jupyter_bokeh` を使います。

````python
import jupyter_bokeh as jbk

from bokeh.models import Slider
from bokeh.layouts import column

fig, (line_sin, line_cos) = plot_demo()
freq_slider = Slider(title="freq", start=1, end=10, step=0.1, value=1)
layout = column(freq_slider, fig)


def update_plot(name, old, new):  # ❷
    x = line_sin.data_source.data["x"]
    line_sin.data_source.data.update(y=np.sin(2 * np.pi * new * x))
    x = line_cos.data_source.data["x"]
    line_cos.data_source.data.update(y=np.cos(2 * np.pi * new * x))


freq_slider.on_change("value", update_plot)  # ❶
jbk.BokehModel(layout)  # ❸
````

### Bokeh サーバー

Bokeh のサーバー機能を使うと、複数の利用者に別々の画面を出せます。サーバーは、利用者のスクリプトを別のプロセスで動かします。スクリプトのあるフォルダで次のコマンドを実行すると、サーバーが立ち上がり、ブラウザで開きます。

```
bokeh serve bokeh_serve_demo.py -show
```

次は、Notebook の `%%writefile` で、コードを `bokeh_serve_demo.py` に書きます。

````python
%%writefile bokeh_serve_demo.py
import numpy as np
from bokeh.io import curdoc
from bokeh.plotting import figure
from bokeh.models import Slider
from bokeh.layouts import column

def plot_demo():
    x = np.linspace(0, 1, 200)    
    fig = figure(width=500, height=300, title="正弦余弦曲線", x_range=(0, 1))
    line_sin = fig.line(x, np.sin(2*np.pi*x), color="red", legend_label="sin", line_width=2, name="sin")
    line_cos = fig.line(x, np.cos(2*np.pi*x), color="green", legend_label="cos", line_width=2, name="cos")
    fig.legend[0].update(location="bottom_center", orientation="horizontal")
    return fig, (line_sin, line_cos)

def serve_demo(doc):
    fig, (line_sin, line_cos) = plot_demo()
    freq_slider = Slider(title="freq", start=1, end=10, step=0.1, value=1)

    def freq_changed(attr, old, new):
        freq = new
        x = line_sin.data_source.data["x"]
        y = np.sin(2*np.pi*freq*x)
        line_sin.data_source.data = dict(x=x, y=y)

        x = line_cos.data_source.data["x"]
        y = np.cos(2*np.pi*freq*x)
        line_cos.data_source.patch({"y":[np.s_[:, y]]})
        
    freq_slider.on_change("value", freq_changed)

    layout = column(freq_slider, fig)
    doc.add_root(layout)
    
serve_demo(curdoc())
````

図のように、ブラウザで新しい窓を開いてアドレスにアクセスするたびに、プログラム全体が新しい環境で動き、新しい `Document` が作られます。Python の `Document` がブラウザに送られ、JavaScript 側にも同じ構造の `Document` ができます。だから、利用者ごとにスライダーの位置が別々で、違う周波数の波形を見られます。

## Bokeh のモデル

Bokeh のクラスは、すべて `Model` から派生しています。`Model` は `HasProps` から派生しています。`HasProps` を継承したクラスは、Bokeh 属性を定義できます。Bokeh 属性には、型、値の検査、既定値、説明などがあります。

次のプログラムは、図のオブジェクト `figure` の派生の並びを出します。

````python
from bokeh.plotting import figure

figure.mro()
````

図では、実線の矢印が Bokeh 属性、点線の矢印が `@property` の属性、灰色の四角がリストや辞書です。たとえば x 軸の `LinearAxis` は `fig.xaxis[0]` でも `fig.below[0]` でも取れます。`xaxis` が `@property`、`below` が Bokeh 属性です。

`Model` の属性を見やすくするために、次のコードは `figure` を JSON にして、ツリーとして表示します。

````python
import json

from IPython.display import display_json

show(fig)

# to_json() は文字列か、Serialized を返す。中身は JSON にできる形に直す。
data = fig.document.to_json()
if hasattr(data, "content"):  # Bokeh 3.9 では Serialized が返ることがある
    data = data.content
if isinstance(data, str):
    data = json.loads(data)


def plain(obj):
    # Buffer など JSON にできない値は、型の名前だけ残す。
    if isinstance(obj, dict):
        return {k: plain(v) for k, v in obj.items()}
    if isinstance(obj, (list, tuple)):
        return [plain(v) for v in obj]
    if isinstance(obj, (str, int, float, bool)) or obj is None:
        return obj
    return f"<{type(obj).__name__}>"


display_json(plain(data)["roots"][0], raw=True)
````

### モデルを探す

図はたくさんのモデルからなる木です。目当てのモデルを探すには、属性をたどる必要があります。`Model` には `references()`、`select()`、`select_one()` があり、まとめて探せます。`select()` は条件に合うもののリスト、`select_one()` は 1 つを返します。2 つ以上合うと `select_one()` は例外を出します。

条件は辞書で書きます。次のプログラムは、`line_width` が `1` のモデルを探します。

````python
fig.references()
````

````python
fig.select(dict(line_width=1))
````

````python
from bokeh.models import Grid

for grid in fig.select(dict(type=Grid)):
    grid.grid_line_dash = "dotted"
````

`bokeh.core.query` には、`IN`、`OR`、`GT`、`LT` などの演算子があります。次は `IN` で `Circle` か `Line` を選びます。

````python
from bokeh.core.query import IN, OR
from bokeh.models import Circle, Line

fig.select(dict(type={IN: [Circle, Line]}))
````

### コールバック関数

`on_change()` で、属性が変わったときに呼ぶ関数を登録できます。関数には、属性名、元の値、新しい値が渡ります。`remove_on_change()` で外せます。次の例は、`line.visible` が `True` から `False` になったときに `on_visible_changed()` を呼びます。

この関数が働くのは、Bokeh サーバーで表示しているときだけです。

````python
def on_visible_changed(name, old, new):
    print(name, old, new)


line.visible = True
line.on_change("visible", on_visible_changed)
line.visible = False
line.remove_on_change("visible", on_visible_changed)
````

図の描画はブラウザの JavaScript が行い、Python のモデルに対応するものが JavaScript 側にもできます。`js_on_change()` で、JavaScript の処理を登録できます。第 1 引数がイベント名、第 2 引数が `CustomJS` です。`change:属性名` か属性名をそのまま書きます。処理の中では `cb_obj` がイベントを起こしたモデルです。`args` でほかのモデルも渡せます。

次の例は、`fig.x_range` の `start` が変わったときに `callback` を呼びます。`console.log()` の結果は、ブラウザの開発者ツールで見てください。

````python
from bokeh.models import CustomJS

fig = figure(width=500, height=300, title="example", x_range=(1, 3))
line = fig.line([1, 2, 3], [4, 6, 5], color="red", legend_label="line1")

callback = CustomJS(
    args=dict(line=line),
    code="""
console.log(line, cb_obj);
""",
)

fig.x_range.js_on_change("start", callback)
show(fig)
````

### レイアウト

`bokeh.layouts` には、`Row`、`Column`、`GridBox` があり、部品を横、縦、格子に並べます。`row()`、`column()`、`grid()` の関数で作れます。`row()` と `column()` は、並べるものをリストか引数で渡します。`grid()` は入れ子のリストか、1 次元のリストと `ncols` で列数を決めます。

次の例では、`row()` に 4 つ、`grid()` に 10 個を 5 列で並べ、`column()` でその 2 つを縦に重ねます。

````python
from bokeh.layouts import row, column, grid, Row, Column, GridBox
from bokeh.palettes import Category10_10
from bokeh.models import Div

layout = column(
    row([Div(width=65, height=50, background="lightgray") for i in range(4)]),
    grid(
        [Div(width=50, height=50, background=color) for color in Category10_10], ncols=5
    ),
)
show(layout)
````

## まとめ

* Bokeh はブラウザで描く。Python はデータと構造を作る。
* Notebook は `output_notebook()`、HTML は `save()`、アプリは Bokeh サーバー。
* 更新は 3 通り。JavaScript(`CustomJS` で書く)、Python(Bokeh サーバーが要る)、`jupyter_bokeh` での試し。
* すべてのクラスは `Model` → `HasProps` の下にあり、Bokeh 属性で型と既定値を持つ。
* モデルは `select()`、`select_one()`、`references()` で探す。道筋は `find_path()`。
* 属性の変化は `on_change()`(Python)と `js_on_change()`(JavaScript)で受ける。
* 部品は `bokeh.models.widgets`、並べ方は `row()`、`column()`、`grid()`。
