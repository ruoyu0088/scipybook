# 用 Panel 做 UI

Panel 是做浏览器里运行的仪表盘 UI 的库，底子是 Bokeh 的部件与布局，在 Notebook 或 Bokeh 服务器里负责画面与 Python 之间的来往。

先读入 HoloViews 与 Panel，再用 `pn.extension()` 初始化 Notebook 用的插件：参数是要读入的插件的名字，这里读入数式用的 katex 与代码显示用的 ace。

- 用 `interact`、`bind`、`param` 做 UI 的 4 种写法
- 傅里叶级数的 UI
- Widget / Pane / Panel 这 3 个部件

```python
import panel as pn
import numpy as np
import holoviews as hv

hv.extension("bokeh")
pn.extension("katex")
```

## UI 的做法

下面 `plot_sin_wave()` 用 4 个参数做出衰减的正弦波，返回 HoloViews 的曲线；接下来要让它这 4 个参数能用部件来改。

```python
def plot_sin_wave(freq, amp, phase, dump):
    t = np.linspace(0, 10, 400)
    y = amp * np.exp(-dump * t) * np.sin(2 * np.pi * freq * t + np.deg2rad(phase))
    return hv.Curve((t, y)).opts(width=600, show_grid=True)
```

### `interact`

`interact` 会自动做出能拖动函数参数的部件，用关键字决定各参数的范围。

给的值的种类不同，部件的种类也不同。

* 元组 `(min, max)`：拖动连续值的滑块。
* `range` 或列表：从固定的值里选的滑块。

```python
ui = pn.interact(
    plot_sin_wave, freq=(1, 10), amp=(1, 10), phase=range(0, 360, 45), dump=(0.0, 1.0)
)
ui
```

看作出来的 UI 里有什么。

* 外层的 `Column` 是整体的容器。
* 里面的 `Column` 是排部件的容器。
* 里面的 `Row` 里放着图的 HoloViews 对象。

```python
print(ui)
```

`Column` 与 `Row` 用 `objects` 嵌套地持有内容。下面的代码把外层改成 `Row`，让部件与图左右并排：因为内容是同一个对象，在新的 UI 里改值，原来的图也会跟着变。

```python
pn.Row(*ui.objects)
```

### 用回调做

`interact()` 方便，但做不了细的调整。复杂的 UI 就自己做出部件与图再连起来，最常用的连法是回调。

❶ 登记部件 `value` 变化时调用的函数，❷ 在那个函数里调用 `plot_sin_wave()`，把结果放进 `fig_panel.object`，换掉图。

```python
freq = pn.widgets.FloatSlider(value=5, start=1, end=10, name="freq")
amp = pn.widgets.FloatSlider(value=1, start=1, end=5, name="amp")
phase = pn.widgets.Select(
    value=90, options=np.arange(0, 360, 45).tolist(), name="phase"
)
dump = pn.widgets.Spinner(value=0.5, step=0.1, name="dump")

ctrl_panel = pn.Column(freq, amp, phase, dump)
fig_panel = pn.pane.HoloViews()


def update_figure(*events):
    fig_panel.object = plot_sin_wave(
        freq.value, amp.value, phase.value, dump.value
    )  # ❷


for widget in [freq, amp, phase, dump]:
    widget.param.watch(update_figure, "value")  # ❶

update_figure()  # ❸
pn.Row(ctrl_panel, fig_panel)
```

### 用 `bind()` 做

回调的代码往往变得复杂，用 `bind()` 就能把函数与部件绑起来，写得短一些。

❶ 用 `bind()` 把 `plot_sin_wave` 与 4 个部件绑起来，部件的值一变函数就被调用，❷ 返回值是 `binded_func` 这个特别的函数，可以直接交给 HoloViews 的 pane。

```python
freq = pn.widgets.FloatSlider(value=5, start=1, end=10, name="freq")
amp = pn.widgets.FloatSlider(value=1, start=1, end=5, name="amp")
phase = pn.widgets.Select(
    value=90, options=np.arange(0, 360, 45).tolist(), name="phase"
)
dump = pn.widgets.Spinner(value=0.5, step=0.1, name="dump")

ctrl_panel = pn.Column(freq, amp, phase, dump)
binded_func = pn.bind(plot_sin_wave, freq, amp, phase, dump)  # ❶
fig_panel = pn.pane.HoloViews(binded_func)  # ❷
pn.Row(ctrl_panel, fig_panel)
```

### 用 `Parameterized` 做

到这里为止都是自己做部件，继承 `param.Parameterized` 就能把参数收进一个类，再从这个定义自动做出 UI；Param 的细节放在这一章最后。

下面的代码做出 `WaveParam`，用 `param.Number` 或 `param.Selector` 决定属性：❶ 做出 `WaveParam` 的对象，❷ 把它传给 `Row()`，部件就自动做出来了，❸ 把 `pars.param` 传给 `bind()`。

```python
import param


class WaveParam(param.Parameterized):
    freq = param.Number(default=5, bounds=(1, 10))
    amp = param.Number(default=1, bounds=(1, 5))
    phase = param.Selector(default=90, objects=np.arange(0, 360, 45).tolist())
    dump = param.Number(default=0.5, step=0.1)


pars = WaveParam()  # ❶
layout = pn.Row(
    pars,  # ❷
    pn.bind(
        plot_sin_wave, *[pars.param[name] for name in ["freq", "amp", "phase", "dump"]]
    ),  # ❸
)
layout
```

做出来的 UI 内容如下。

```python
print(layout)
```

更新图的函数也能写成类的方法：❶ 用 `depends()` 装饰方法，就与属性绑在一起，不用 `bind()`，❷ 把 `pars` 与 `pars.plot` 传给 `Row`，UI 与图就在一起了。

```python
class WaveParam(param.Parameterized):
    freq = param.Number(default=5, bounds=(1, 10))
    amp = param.Number(default=1, bounds=(1, 5))
    phase = param.Selector(default=90, objects=np.arange(0, 360, 45).tolist())
    dump = param.Number(default=0.5, step=0.1)

    @pn.depends("freq", "amp", "phase", "dump")  # ❶
    def plot(self):
        return plot_sin_wave(self.freq, self.amp, self.phase, self.dump)


pars = WaveParam()
layout = pn.Row(pars, pars.plot)  # ❷
layout
```

## 傅里叶级数的 UI

这个例子用 UI 看各种函数的傅里叶级数波形：用 `function` 下拉框选函数、用 `order` 滑块选次数，展开的式子显示在下面、曲线显示在右边。

先看计算的部分：`calc_fourier(func, order)` 把函数 `func` 展开成傅里叶级数，返回前 `order` 项，返回值有 2 个，LaTeX 的式子与这些项的 `Curve`；因为要反复调用，这里用 `functools.lru_cache()` 记住结果。

```python
from functools import lru_cache
import numpy as np
from sympy import Piecewise, S, Symbol, And, fourier_series, lambdify, latex, pi


@lru_cache()
def calc_fourier(func, order):
    x = Symbol("x")
    fs = fourier_series(func, (x, 0, 2))
    fs_t = fs.truncate(n=order)
    fs_lambda = lambdify([x], fs_t)
    x_ = np.linspace(0, 4, 200)
    y_ = fs_lambda(x_)
    return f"$${latex(fs_t)}$$", hv.Curve((x_, y_))


x = Symbol("x")

functions = {
    "step": Piecewise((1, x < 1), (0, True)),
    "triangle": Piecewise((x, x < 1), (2 - x, True)),
    "pulse": Piecewise((1, And(S(2) / 3 < x, x < S(4) / 3)), (0, True)),
    "trapezoid": Piecewise(
        (S(3) / 2 * x, x < S(2) / 3), (1, x < S(4) / 3), (3 - S(3) / 2 * x, True)
    ),
    "sawtooth": x,
}
```

这是阶梯函数的傅里叶级数前 5 项的式子与图。

```python
from IPython.display import display_latex

formula, curve = calc_fourier(functions["step"], 5)
display_latex(formula, raw=True)
curve.opts(width=600, height=300)
```

### 做成响应式

做出 `calc_fourier()` 的 UI，部件都收进 `FourierSeriesViewer` 类里。

❶ 做出两个部件：下拉框与整数的滑块，❷ 用 `bind()` 把 `calc_fourier` 与两个部件绑起来。❸ 返回值有 2 个，不能直接交给 pane，于是给 `rx()` 的对象加 `[]`，把第 1 个交给 `LaTeX`、第 2 个交给 `HoloViews`，❹ 最后把 UI 的布局拼起来。

```python
class FourierSeriesViewer:
    def __init__(self, functions):
        self.functions = functions
        self.func_name = pn.widgets.Select(name="function", options=self.functions)  # ❶
        self.order = pn.widgets.IntSlider(
            name="order", value=5, start=1, end=10, step=1
        )
        func = pn.bind(calc_fourier, self.func_name, self.order)  # ❷
        self.latex_pane = pn.pane.LaTeX(func.rx()[0])  # ❸
        self.curve_pane = pn.pane.HoloViews(func.rx()[1].opts(width=600, height=300))
        self.layout = pn.Column(  # ❹
            pn.Row(pn.Column(self.func_name, self.order), self.curve_pane),
            self.latex_pane,
        )


fsv = FourierSeriesViewer(functions)
fsv.layout
```

`rx()` 把函数或表达式变成“响应式”的：响应式的对象在输入变化时会重新计算输出。`bind()` 只管参数与返回值，而 `rx()` 能把整个 Python 计算都变成响应式的。

下面 `func()` 返回 2 个值，`pn.bind(func, x=slider).rx()[1]` 就把第 2 个返回值与滑块绑起来。

```python
slider = pn.widgets.FloatSlider(name="x", value=5, start=1, end=10)


def func(x):
    return [x + 1, x * 10]


pn.bind(func, x=slider).rx()[1]
```

### 用回调做(参数版)

把一个 `Parameter` 放进 `param.Parameterized` 的派生类里，部件的创建与事件的处理都变简单，下面 `FourierSeriesViewer2` 有 `func` 与 `order` 两个 `Parameter`。

`__init__()` 里 ❶ 用 `self.param.func.objects` 放进候选的值。❷ 出 UI 之前就要决定 `latex_pane` 与 `curve_pane` 的内容，因此给 `func` 放进第一个值时 `calc()` 就会跑起来。

```python
import param


class FourierSeriesViewer2(param.Parameterized):
    func = param.Selector()
    order = param.Integer(default=5, bounds=(1, 10))

    def __init__(self, functions, **kw):
        super().__init__(**kw)
        self.param.func.objects = functions  # ❶

        self.latex_pane = pn.pane.LaTeX()
        self.curve_pane = pn.pane.HoloViews()
        self.layout = pn.Column(pn.Row(self, self.curve_pane), self.latex_pane)

        self.func = next(iter(functions.values()))  # ❷

    @param.depends("func", "order", watch=True)  # ❸
    def calc(self):
        order = self.order
        formula, curve = calc_fourier(self.func, order)
        self.latex_pane.object = formula
        self.curve_pane.object = curve.opts(width=500, show_grid=True)


fsv2 = FourierSeriesViewer2(functions)
fsv2.layout.servable()
```

## 用 Param 定义参数

Panel 的类从 `param.Parameterized` 派生，属性全都是 Parameter，这一节说明 Param 的用法。

Python 里用 `__init__()` 初始化属性，但那只能存值；用 Param 定义属性，就能连类型、值的范围与说明一起带上。

### Parameter 属性

持有 Parameter 的类从 `param.Parameterized` 派生，属性用 `param.Parameter` 的派生类来定义。

下面的例子里 ❶ `Calculator` 从 `param.Parameterized` 派生，有 `op1`、`op2`、`operator`、`result` 4 个属性。❷ `op1` 与 `op2` 是 `Number`，用 `default`、`step`、`bounds`、`label` 决定默认值、步长、范围与显示名。❸ `operator` 是 `Selector`，可选的值排在 `objects` 里。

```python
import param


class Calculator(param.Parameterized):  # ❶
    op1 = param.Number(default=2.0, step=0.1, bounds=(-10, 10), label="演算数1")  # ❷
    op2 = param.Number(default=3.0, step=0.1, bounds=(-10, 10), label="演算数2")
    operator = param.Selector(objects=list("+-*/"), label="演算子")  # ❸
    result = param.String(label="結果")  # ❹
```

Parameter 的类都从 `param.Parameter` 派生，常用的如下。

| 类型 | 说明 | 主要参数 |
|:--|:--|:--|
| `String`  | 字符串  | `regex`、`allow_None` |
| `Number`  | 数值  | `bounds`、`step` |
| `Boolean`  | 真假值  | |
| `Filename`  | 文件的路径 | `search_paths` |
| `List` | 列表 | `class_` |
| `Tuple` | 元组 | `length` |
| `DataFrame` | pandas 的 `DataFrame` | |
| `ListSelector` | 从一览里选 | `objects` |
| `Selector` | 选 1 个 | `objects` |
| `Range` | 范围 | `bounds` |

```python
print_subclasses(param.parameterized.Parameter, as_html=True, height=300)
```

### 回调

给 Parameter 放值时，能调用决定的函数。

❶ 用 `c1.param.watch()` 给 `c1` 的 Parameter 登记函数：第 1 个参数是函数、第 2 个是属性名或它的列表；`onlychanged=False` 时每次赋值都调用，`True` 时只在值变化时调用，返回值是 `Watcher`。

❷ 传进来的参数个数不定，函数因此用 `*events` 来接；每个参数是 `Event`，带着 `name`、`old`、`new`。

```python
def callback(*events):  # ❷
    for event in events:
        print(
            f"obj: {event.obj.name if event.obj is not None else None}, ",
            f"what: {event.what}, name: {event.name}, "
            f"old: {event.old}, new: {event.new}, type: {event.type}",
        )
    print()


c1 = Calculator(op1=0, op2=0, name="c1")
watch = c1.param.watch(callback, ["op1", "op2"], onlychanged=False)  # ❶
print("c1.op1 = 5")
c1.op1 = 5  # ❸

print("c1.update(op1=3, op2=6)")
c1.param.update(op1=3, op2=6)  # ❹

print("batch_watch() for loop")
with param.parameterized.batch_call_watchers(c1):  # ❺
    for i in range(8):
        c1.op1 = i
        c1.op2 = i + 1

c1.param.unwatch(watch)  # ❻
print("c1.op1 = 6")
c1.op1 = 6
```

用 `watch()` 的 `what` 也能接收值以外的信息的变化，下面的程序在 `op2` 的范围变化时调用函数。

```python
c1.param.watch(callback, "op2", what="bounds")
c1.param.op2.bounds = (0, 1000)
```

定义类时用 `param.depends()` 能把方法变成回调。

❶ `Calculator2` 继承 `Calculator`，`__init__()` 接收 `**kw`，❷ 为了正确放进属性，要在别的初始化之前先调用父类的 `__init__()`，❸ 用 `self.param.trigger()` 触发 `operator` 的事件。❹ 用 `param.depends()` 把 `run()` 与 `op1`、`op2`、`operator` 绑起来，只有 `watch=True` 时才会登记成回调。

```python
class Calculator2(Calculator):  # ❶
    def __init__(self, **kw):
        super().__init__(**kw)  # ❷
        self.param.trigger("operator")  # ❸

    @param.depends("op1", "op2", "operator", watch=True)  # ❹
    def run(self):
        import operator as op

        funcs = {"+": op.add, "-": op.sub, "*": op.mul, "/": op.truediv}
        result = funcs[self.operator](self.op1, self.op2)
        self.result = f"{self.op1:4.1f} {self.operator} {self.op2:4.2f} = {result:4.2f}"


c2 = Calculator2()
print(c2.result)  # ❺
c2.op1 = 5
c2.operator = "*"
print(c2.result)
```

### 自动做 UI

`Parameterized` 的派生类常用来自动做 UI：先读入 Panel，用 `extension()` 初始化 Notebook 的插件。

## 小结

* Panel 是架在 Bokeh 上的 UI 库，用 `pn.extension()` 初始化。
* UI 有 4 种做法：`interact()`(自动)、回调、`bind()`、`param.Parameterized`(自动)。
* `bind()` 把函数与部件绑起来，用 `rx()` 连计算途中也能做成响应式。
* 部件分 Widget、Pane、Panel 3 层，分别在 `panel.widgets`、`panel.pane`、`panel.layout` 里。
* Param 的 Parameter 是带着类型、范围与说明的属性，用 `watch()` 与 `depends()` 接收变化。
* 用 `panel(obj)` 能从 `Parameterized` 的对象自动做出 UI。
